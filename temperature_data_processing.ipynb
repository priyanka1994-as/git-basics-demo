```python
import numpy as np

# Create NumPy array
temps_celsius = np.array([22, 25, 28, 24, 26])

# Convert to Fahrenheit
temps_fahrenheit = temps_celsius * 1.8 + 32

# Calculate average (rounded to 1 decimal place)
average_fahrenheit = np.round(np.mean(temps_fahrenheit), 1)

# Print results
print("Celsius:", temps_celsius)
print("Fahrenheit:", temps_fahrenheit)
print("Average Fahrenheit:", average_fahrenheit)
```

```python
import numpy as np

# Create the array
scores = np.array([85, 90, 78, 92, 88, 76, 95, 82, 89, 91, 87, 84])

# Shape of the array
print("Shape:", scores.shape)

# Total number of elements
print("Total elements:", scores.size)

# Highest score
highest = np.max(scores)
print("Highest score:", highest)

# Lowest score
lowest = np.min(scores)
print("Lowest score:", lowest)

# Range
range_value = highest - lowest
print("Range:", range_value)
```

```python
import numpy as np
import time

# Create NumPy array
np_array = np.arange(1, 50001)

# Create Python list
py_list = list(range(1, 50001))

# NumPy sum timing
start_np = time.time()
np_sum = np.sum(np_array)
end_np = time.time()
np_time = end_np - start_np

# Python sum timing
start_py = time.time()
py_sum = sum(py_list)
end_py = time.time()
py_time = end_py - start_py

# Calculate speed difference
speed_difference = py_time / np_time

# Print results
print("NumPy sum:", np_sum)
print("Python sum:", py_sum)
print("NumPy time:", format(np_time, ".4f"), "seconds")
print("Python time:", format(py_time, ".4f"), "seconds")
print("NumPy is", format(speed_difference, ".1f") + "x faster")
```
