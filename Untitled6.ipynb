```python
import pandas as pd
import numpy as np

# Original DataFrame
df = pd.DataFrame({
    'customer_id': [1, 2, 3, 4, 2, 5],
    'age':         [25, np.nan, 35, np.nan, np.nan, 40],
    'city':        ['Mumbai', 'Delhi', np.nan, 'Mumbai', 'Delhi', np.nan],
    'gender':      ['Male', 'Female', 'Female', 'Male', 'Female', 'Male'],
    'annual_income': [40000, 80000, 60000, 120000, 80000, 95000]
})

# 1. Impute missing values in 'age' with median
median_age = df['age'].median()
df['age'].fillna(median_age, inplace=True)

# 2. Impute missing values in 'city' with mode
mode_city = df['city'].mode()[0]
df['city'].fillna(mode_city, inplace=True)

# 3. Remove duplicate rows
df = df.drop_duplicates()

# Display cleaned DataFrame
print(df)
```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# Assuming df is already cleaned from Task 1

# 1. One-Hot Encoding for 'city'
df = pd.get_dummies(df, columns=['city'])

# 2. Label Encoding for 'gender'
le = LabelEncoder()
df['gender'] = le.fit_transform(df['gender'])

# Display final DataFrame
print(df)
```

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler

# Create copies so original data is preserved
df_minmax = df.copy()
df_standard = df.copy()

# Columns to scale
cols = ['age', 'annual_income']

# 1. Min-Max Scaling
minmax_scaler = MinMaxScaler()
df_minmax[cols] = minmax_scaler.fit_transform(df_minmax[cols])

print("Min-Max Scaled Data:")
print(df_minmax)

# 2. Standardisation
standard_scaler = StandardScaler()
df_standard[cols] = standard_scaler.fit_transform(df_standard[cols])

print("\nStandardised Data:")
print(df_standard)
```

🧾 Explanation (2–3 sentences):

Use Min-Max Scaling when your data needs to be in a fixed range (e.g., neural networks or distance-based models like KNN). Use Standardisation when the data may contain outliers or when algorithms assume normally distributed data (e.g., linear regression, logistic regression). Standardisation is generally more robust when feature scales vary widely.
