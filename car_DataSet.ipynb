```python
Task 1 — Explore and Identify Issues
```

```python
import pandas as pd

# Load dataset
df = pd.read_csv("used_cars.csv")

# Basic exploration
print("Shape:", df.shape)
print("\nInfo:")
print(df.info())
print("\nDescribe:")
print(df.describe(include='all'))
```

```python
Task 2 — Clean the Data
```

```python
# 1. Drop rows where target (selling_price) is null
df = df.dropna(subset=['selling_price'])

# 2. Handle missing values in input features
# (Example: fill numeric columns with median)
numeric_cols = df.select_dtypes(include=['float64', 'int64']).columns

for col in numeric_cols:
    if col != 'selling_price':
        df[col] = df[col].fillna(df[col].median())

# 3. Clean 'brand' column
df['brand'] = df['brand'].astype(str).str.strip().str.lower()

# 4. Extract numeric values from 'mileage'
# Example: "18 kmpl" → 18
df['mileage'] = df['mileage'].astype(str).str.extract('(\d+\.?\d*)')
df['mileage'] = df['mileage'].astype(float)

# Fill missing mileage after extraction
df['mileage'] = df['mileage'].fillna(df['mileage'].median())

# 5. Remove duplicates
df = df.drop_duplicates()

# Final check
print("Cleaned Shape:", df.shape)
print(df.head())
```

```python
Task 3 — Compute Baseline MAE
```

```python
from sklearn.metrics import mean_absolute_error

# Calculate mean selling price
mean_price = df['selling_price'].mean()

# Create predictions (same value for all rows)
y_true = df['selling_price']
y_pred = [mean_price] * len(df)

# Compute MAE
mae = mean_absolute_error(y_true, y_pred)

print("Baseline Mean Price:", mean_price)
print("Baseline MAE:", mae)
```
